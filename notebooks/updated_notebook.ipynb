```python
df = df.drop(['student_id','hsc_board','hsc_stream','mba_percentage','specialization','certifications_count','technical_skills_score','soft_skills_score','aptitude_score','work_experience_months','leadership_roles','extracurricular_activities','salary_lpa'],axis = 1)
```

```python
df = df.drop(['ssc_board'], axis = 1)
```

```python
df.columns
```

```python
df.head()
```

```python
df = df.replace(['Male' , 'Female' , 'Other'] , [0,1,2])
```

```python
df.head()
```

```python
df.tail()
```

```python
df['degree_field'].unique()
```

```python
df['degree_field'].isnull().sum()
```

```python
 df = df[~df['degree_field'].isin([2 ,'2'])]
```

```python
df['degree_field'].unique()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
x = df.drop(['placed'],axis = 1)
y = df['placed']
```

```python
x = x.rename(columns = {'ssc_percentage' : 'tenth_percentage' , 'hsc_percentage' : 'twelfth_percentage'})
```

```python
x.columns
```

```python
df['degree_field'].unique()
```

```python
 x = x.drop(['backlogs'] , axis = 1)
```

```python
 x = x.drop(['communication_score'] , axis = 1)
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x , y ,test_size = 0.2  , random_state = 42)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
categorical_cols = [
    'city_tier',
    'degree_field'
]
```

```python
preprocessor = ColumnTransformer(transformers=[('cat',OneHotEncoder(handle_unknown='ignore'),categorical_cols)],remainder='passthrough')
```

```python
pipe = Pipeline([('preprocessor', preprocessor),('model', LogisticRegression())])
```

```python
pipe.fit(x_train , y_train)
```

```python
y_pred = pipe.predict(x_test)
```

```python
pipe.score(x_test , y_test)
```

```python
import joblib

joblib.dump(pipe, "placement_predictor.pkl")
```
